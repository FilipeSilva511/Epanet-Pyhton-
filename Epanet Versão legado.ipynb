## Alterando a Rugosidade

```python
# Iterando sobre os IDs fornecidos
count_item = 0
for id in lista_ids:
    # Garantir que o id seja tratado como string para corresponder à chave do dicionário
    str_id = str(id)  # Convertendo o ID para string
    
    # Verificar se o ID existe no dicionário
    if str_id in dict_ids:
        # Obter o índice correspondente ao ID
        index = dict_ids[str_id]
        
        # Recupera o valor atual de rugosidade
        current_roughness = float(d.getLinkRoughnessCoeff(index))
        
        # Verificar se o valor de rugosidade é um número (float ou int)
        
        if isinstance(current_roughness, float) or isinstance(current_roughness, int):
            count_item += 1
            # Subtrair 10 unidades da rugosidade e definir o novo valor
            novo_valor = current_roughness - 10
        
            # Ajustar a rugosidade com o novo valor
            d.setLinkRoughnessCoeff(index, novo_valor)
            print(f"Rugosidade do ID {id} (Índice {index}) ajustada para {novo_valor}, valor antigo: {current_roughness}, valor novo: {novo_valor}")
        else:
            # Caso o valor de rugosidade não seja numérico, exibir uma mensagem de erro
            print(current_roughness)
            print(f"Erro ao recuperar a rugosidade para o ID {id}. Valor de rugosidade inválido.")
    else:
        # Caso o ID não seja encontrado no dicionário, exibir uma mensagem de erro
        print(f"ID {id} não encontrado no dicionário.")
```

## Alterando a demanda dos Nós

```python
# Garante que os IDs estão no formato correto (str para comparação)
lista_nos = [str(id) for id in lista_nos]

# Criar o dicionário de IDs (nome -> índice)
dict_ids = {nome: index for nome, index in zip(d.getNodeNameID(), d.getNodeIndex())}

# Inicializar as variáveis de soma
soma_demanda_antiga = 0
soma_demanda_nova = 0
soma_reduzida = 0

# Percorrer os nós da lista
for id in lista_nos:
    if id in dict_ids:
        index = dict_ids[id]

        # Obter todas as demandas do nó (em formato de dicionário)
        demandas_no = d.getNodeBaseDemands(index)

        # Somar e ajustar cada demanda associada ao nó
        for demanda_atual in demandas_no.items():
            demanda_atual = float(list(d.getNodeBaseDemands(index).values())[0])  # Garantir que é numérico

            # Reduzir a demanda em 5%
            nova_demanda = demanda_atual - (demanda_atual*0.025)
            d.setNodeBaseDemands(index,  nova_demanda)

            # Atualizar as somas
            soma_demanda_antiga += demanda_atual
            soma_demanda_nova += nova_demanda
            soma_reduzida += (demanda_atual - nova_demanda)

            print(f"Nó {id} (Índice {index}) ajustada para {nova_demanda}, valor antigo: {demanda_atual}")
    else:
        print(f"ID {id} não encontrado no dicionário.")

# Exibir o resumo final
print("\n🔎 Resumo Final:")
print(f"Soma total das demandas antigas: {soma_demanda_antiga:.2f}")
print(f"Soma total das demandas novas: {soma_demanda_nova:.2f}")
print(f"Redução total aplicada: {soma_reduzida:.2f}")
```
